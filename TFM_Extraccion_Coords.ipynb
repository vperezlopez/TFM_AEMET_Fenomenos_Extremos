```python
import pandas as pd
import requests
```

```python
URL_FICHERO_API_KEY = 'api_key.txt'
with open(URL_FICHERO_API_KEY, 'r') as fichero_api_key:
    API_KEY = fichero_api_key.read().strip()
```

```python
URL_ESTACIONES = 'https://opendata.aemet.es/opendata/api/valores/climatologicos/inventarioestaciones/todasestaciones'

URL_DATOS = 'datos'

headers = {
        'accept': 'application/json',
        'api_key': API_KEY
    }
```

```python
response = requests.get(URL_ESTACIONES, headers=headers).json()

url_estaciones = response[URL_DATOS]
response_estaciones = requests.get(url_estaciones, headers=headers)

df_coords = pd.DataFrame(response_estaciones.json())
```

```python
display(df_coords.sample(20))
```

```python
df_coords.to_csv('Inventario_Estaciones_AEMET.csv', sep = ';', index = False)
```
