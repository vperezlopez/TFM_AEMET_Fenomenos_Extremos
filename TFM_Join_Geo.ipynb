```python
import pandas as pd
```

```python
PATH_VAL_CLIM = 'Valores_Climatologicos_1970_2024.csv'
```

```python
df = pd.read_csv(PATH_VAL_CLIM, sep=';', dtype=str)
```

```python
display(df.sample(20))
```

```python
PATH_ESTACIONES = 'Inventario_Estaciones_AEMET.csv'
```

```python
df_estaciones = pd.read_csv(PATH_ESTACIONES, sep=';', dtype=str)
```

```python
display(df_estaciones.sample(20))
```

```python
def procesar_coordenada(coord_str: str) -> float:
    """
    Convierte coordenadas en formato DMS compacto a formato decimal (float).
    """

    direccion = coord_str[-1] # Último carácter, la letra (N, S, E, W)
    coord = coord_str[:-1] # Todos los demás dígitos

    # Formato DMS: DDMMSS / DDDMMSS
    grados = int(coord[:-4])    # Los primeros N - 4 dígitos (DD o DDD)
    minutos = int(coord[-4:-2]) # Tercer y cuarto dígitos empezando por el final (MM)
    segundos = int(coord[-2:])  # Últimos dos dígitos (SS)

    coord_float = grados + minutos / 60 + segundos / 3600

    # Sur y oeste se codifican como negativos
    if direccion == 'S' or direccion == 'W':
        coord_float = -coord_float

    return coord_float
```

```python
LON = 'lon'
LAT = 'lat'
```

```python
df_estaciones[LON] = df_estaciones['longitud'].apply(procesar_coordenada)
df_estaciones[LAT] = df_estaciones['latitud'].apply(procesar_coordenada)
```

```python
col_join = 'indicativo'

cols_coords = [
    LON,
    LAT
]

df_merged = df.merge(
    df_estaciones[[col_join] + cols_coords],
    on = col_join,
    how="left"
)
```

```python
display(df_merged.sample(20))
```

```python
df_merged.to_csv('Valores_Climatologicos_1970_2024_con_Coordenadas.csv', sep = ';', index = False)
```
