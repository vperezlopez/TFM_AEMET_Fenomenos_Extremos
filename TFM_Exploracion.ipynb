```python
import pandas as pd
import numpy as np

import geopandas as gpd

from loader import cargar_dataset_limpio
```

```python
df = cargar_dataset_limpio()
```

```python
daily = df.groupby('fecha', as_index=False).agg(
    tmax_mean=('tmax','mean'),
    prec_sum=('prec','sum'),
    racha_max=('racha','max'),
    estaciones=('indicativo','nunique')
)
daily['tmax_mm30'] = daily['tmax_mean'].rolling(30, min_periods=10).mean()
```
