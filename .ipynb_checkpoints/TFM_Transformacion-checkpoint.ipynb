```python
import pandas as pd
import numpy as np
```

```python
PATH_VAL_CLIM = 'Valores_Climatologicos_1970_2024.csv'
```

```python
COLS_NOMBRES = [
    'indicativo',
    'nombre',
    'provincia',
]

COLS_INTS = [
    'altitud'
]

COLS_FLOATS = [
    'tmed',
    'prec',
    'tmin',
    'tmax',
    'velmedia',
    'sol',
    'presMax',
    'presMin',
    'hrMedia',
    'dir',
    'racha',
    'hrMax',
    'hrMin'
]

COLS_NUMS = COLS_INTS + COLS_FLOATS

COL_FECHA = 'fecha'
COLS_HORAS = [
    'horaPresMax',
    'horaPresMin',
    'horatmin',
    'horatmax',
    'horaracha',
    'horaHrMax',
    'horaHrMin'
]
```

```python
# dtypes_dict = {nombre: str for nombre in COLS_NOMBRES} | {num: float for num in COLS_NUMS}
test_dict = {decimal: 'Float64' for decimal in COLS_FLOATS if decimal != 'prec'}
test_dict
```

```python
df = pd.read_csv(PATH_VAL_CLIM,
                 sep=';', decimal = ',',
                 dtype = {nombre: 'string' for nombre in COLS_NOMBRES} |
                         {entero: 'Int64' for entero in COLS_INTS} |
                         {otros: 'object' for otros in COLS_FLOATS + COLS_HORAS},
                         # {decimal: 'Float64' for decimal in COLS_FLOATS if decimal != 'prec'},
                 parse_dates = [COL_FECHA])
```

```python
display(df)
```

```python
df_prep = df.copy(deep = True)
```

```python
COL_PREC = 'prec'
# df_prep.loc[df[COL_PREC] == 'Ip', COL_PREC] = 0
df_prep = df_prep[df_prep[COL_PREC] != 'Acum']
df_prep[COL_PREC] = df_prep[COL_PREC].replace('Ip', 0)
```

```python
df_prep[COLS_FLOATS] = df_prep[COLS_FLOATS].replace(',', '.', regex=True).apply(pd.to_numeric, errors = 'raise')
```

```python
df_prep.dtypes
```

```python
# PENDIENTE: LIMPIEZA Y CONVERSIÓN DE HORAS
# MIRAR DIR: 88 y 99
```

```python
cols_horas = [
    'horaPresMax',
    'horaPresMin',
    'horatmin',
    'horatmax',
    'horaracha',
    'horaHrMax',
    'horaHrMin'
]

df_prep[cols_horas].astype(str).values

# sorted(list(set(df_prep[cols_horas].astype(str).values)))[-5:]

lista_vals_horas = df_prep[cols_horas].astype(str).values

lista_vals_horas_flat = [val for sublista in lista_vals_horas for val in sublista]
```

```python
# Añadimos el sufijo :00 a las columnas cuyos valores carecen de él
cols_sufijo = ['horaPresMax', 'horaPresMin']
df_prep[cols_sufijo] = df_prep[cols_sufijo].where(
    df_prep[cols_sufijo].isna() | (df_prep[cols_sufijo] == 'Varias'),
    df_prep[cols_sufijo] + ':00'
)
```

```python
# Procesamiento de horas
CATEGORIAS = ['Desconocida',
              'Varias',
              'Válida',
              'Inválida'] # Valor por defecto si no cuadra en ninguna de las demás

for col in cols_horas:

    filtro_formato = df_prep[col].str.match(r'^\d{1,2}:\d{2}$', na=False) # Comprueba que formato HH:MM
    df_hhmm = df_prep.loc[filtro_formato, col].str.split(":", expand=True).astype(int) # Df con dos columnas: HH y MM
    filtro_rango = (df_hhmm[0].between(0, 23)) & (df_hhmm[1].between(0, 59)) # Comprueba HH entre 0-23; MM entre 0-59 
    filtro_hhmm = filtro_rango.reindex(df_prep.index, fill_value = False) # Para mismo tamaño que original
    
    # Anotamos anomalías en la hora
    filtros = [
        df_prep[col].isna(),
        df_prep[col] == 'Varias',
        filtro_hhmm
    ]

    df_prep[col + '_tipo'] = np.select(filtros, CATEGORIAS[:-1], default = CATEGORIAS[-1])
    df_prep[col + '_tipo'] = df_prep[col + '_tipo'].astype('category')

    # Pasamos las horas a tipo time
    df_prep[col + '_limpia'] = pd.to_datetime(
        df_prep[col], format = '%H:%M', errors = 'coerce'
    ).dt.time

    # Creamos columna datetime respectiva
    df_prep['fecha' + col + '_limpia'] = pd.to_datetime(
        df_prep[COL_FECHA].dt.date.astype(str) + ' ' + df_prep[col].astype(str),
        format = '%Y-%m-%d %H:%M',
        errors = 'coerce'
    )
    
```

```python
# Procesamos dirección del viento
dict_dir = {
    88: 'Desconocida',
    99: 'Varias'
}
dir_por_defecto = 'Válida'

df_prep['dir_tipo'] = df_prep['dir'].map(dict_dir).fillna(dir_por_defecto)
```

```python
df_prep.to_csv('Valores_Climatologicos_1970_2024_Limpios.csv', sep = ';', index = False)
```

```python
# Reemplazamos los valores superiores a 24:00 cuyo valor real se puede deducir
cols_horas_reemplazar = ['horatmin', 'horatmax']
dict_horas_reemplazar = {'24:00': '00:00', '24:10': '00:10'}
df_prep[cols_horas_reemplazar] = df_prep[cols_horas_reemplazar].replace(dict_horas_reemplazar)
```

```python
# Filtramos aquellas horas superiores a 24:00 y las sustituimos por nan
mask = (
    df_prep['horaracha'].notna()
    & (df_prep['horaracha'] != 'Varias')
    & pd.to_timedelta(df_prep['horaracha'] + ':00', errors='coerce').notna()
    & (pd.to_timedelta(df_prep['horaracha'] + ':00', errors='coerce') > pd.Timedelta(hours=24))
)

df_prep.loc[mask, 'horaracha'] = np.nan
```

```python
# Para cada variable hora, creamos una columna binaria que indica si su valor era 'Varias'
for col in cols_horas:
    df_prep[col + '_Varias'] = df_prep[col] == 'Varias'

# Cambiamos los 'Varias' por nan
df_prep[cols_horas] = df_prep[cols_horas].replace('Varias', np.nan)
```

```python
COL_FECHA = 'fecha'

for col in cols_horas:
    col_nueva = 'fecha' + col

    # Guardar copia para inspección
    originales = df_prep[col].copy()

    # Intentar convertir y guardar en la nueva columna
    df_prep[col_nueva] = pd.to_datetime(
        df_prep[COL_FECHA].dt.strftime('%Y-%m-%d') + ' ' + df_prep[col].astype(str),
        format='%Y-%m-%d %H:%M',
        errors='coerce'
    )

    # Detectar los que fallaron (no NaN originales pero sí NaN convertidos)
    mask_invalidos = df_prep[col_nueva].isna() & originales.notna()
    if mask_invalidos.any():
        print(f"\nColumna: {col}")
        print(originales[mask_invalidos].unique())
```
